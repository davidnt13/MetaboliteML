```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
for dataset_filename in ['../Acetylcholinesterase_CHEMBL220_NP.csv', 
                         '../Monoamine_oxidase_A_CHEMBL1951_NP.csv', 
                         '../Protein-tyrosine_phosphatase_1B_CHEMBL335_NP.csv', 
                        ]:
    df = pd.read_csv(dataset_filename)\
            .set_index('ChEMBL_ID', verify_integrity = True)
    X = df.drop('pIC50', axis = 1)
    y = df['pIC50']
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=1)
    dfHoldOut = X_train.join(y_train)
    dfHoldOut = dfHoldOut.drop('natural_product', axis = 1)
    dfHoldOut.to_csv("Test_" + dataset_filename.removeprefix("../"))
```
